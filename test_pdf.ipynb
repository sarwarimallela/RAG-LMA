```python

import pymupdf

path = "data/try2.pdf"

doc = pymupdf.open(path)
print("Pages:", len(doc))

for i in range(min(3, len(doc))):
    try:
        text = doc[i].get_text(sort=True)
        print(f"\n--- PAGE {i + 1} ---")
        print(text[:1000] if text.strip() else "No text extracted")
    except Exception as e:
        print(f"Page {i + 1} failed: {e}")
```
